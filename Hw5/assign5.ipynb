```python
#PART 1!
#Function that reads the text file and turns it into a vertical database
def readFile(file_name):
    text_file = open(file_name, "r")
    lines = text_file.readlines()
    #print(lines)
    possible_entries = set()
    data_list = []
    #Converts each line from the text file into a set and unionizes them
    #Also creates a list of lists of data from the text file
    count = 1
    for i in lines:
        row_data = i.strip().split()
        possible_entries = possible_entries.union(set(row_data))
        data_list.append((count, row_data))
        count = count + 1

    #Converts the set back into list
    possible_entries = list(possible_entries)

    #Creates a list of tuples in the form (possible entry, set of row numbers the entry occurs in)
    end_data = []
    for j in range(len(possible_entries)):
        record_set = set()
        for k in range(len(data_list)):
            if possible_entries[j] in data_list[k][1]:
                record_set.add(data_list[k][0])
        entry_set = {possible_entries[j]}
        tuple=[entry_set, record_set]
        end_data.append(tuple)
    
    #Sorts the end result by the first entry in each tuple
    end_data.sort()

    #Closes the text file
    text_file.close()

    return end_data

P = readFile("example.txt")
print(P)
```

```python
#PART 1!
#P is the set of potential itemsets
#minsup is the minimum support
#C is the set of closed itemsets
#Algorithm CHARM (9.2)
def charm(P, minsup, C):
    #STEP 1
    P.sort(key=lambda item: len(item[1]))
    #STEP 2
    for i, [X_i, tX_i] in enumerate(P):
        #STEP 3
        P_i = []
        #STEP 4
        for j, [X_j, tX_j] in enumerate(P):
            if j>i:
                #STEP 5
                X_ij = set(X_i).union(set(X_j))
                #STEP 6
                tX_ij = tX_i.intersection(tX_j)
                #STEP 7
                if len(tX_ij) >= minsup:
                    #STEP 8
                    if tX_i == tX_j:
                        #STEP 9
                        for k in range(len(P_i)):
                            if set(X_i).issubset(P_i[k][0]):
                                P_i[k][0] = P_i[k][0].union(X_ij)
                        P[i][0] = X_ij
                        X_i = X_ij
                        #STEP 10
                        P.pop(j)
                    #STEP 11 & 12
                    elif tX_i.issubset(tX_j):
                        #STEP 13
                        for k in range(len(P_i)):
                            if set(X_i).issubset(P_i[k][0]):
                                P_i[k][0] = P_i[k][0].union(X_ij)
                        P[i][0] = X_ij
                        X_i = X_ij
                    #STEP 14
                    else:
                        #STEP 15
                        P_i.append([X_ij, tX_ij])
        #STEP 16
        if P_i != []:
            charm(P_i, minsup, C)
        #STEP 17
        if not any(set(X_i).issubset(Z) and tX_i == t_z for Z, t_z in C) and len(tX_i)>=minsup:
            #STEP 18
            C.append([set(X_i), tX_i])
```

```python
#PART 2
import numpy as np
from sklearn import datasets
from sklearn.metrics import normalized_mutual_info_score #CSCI 4390 students instructed to use
from scipy.stats import multivariate_normal #Instructed to use
from scipy.special import logsumexp #Instructed to use

#Loading in the iris dataset
iris = datasets.load_iris()
X = iris.data[:, :4]  #Take only the first four features.
y_true = iris.target

#Initialization of variables (STEPS 1 through 4)
k = 3  #Number of clusters which we are given
epsilon = 1e-4  #Convergence threshold

#Expectation-Maximization (EM) Algorithm
def EM_algorithm(X, k, epsilon):
    #STEPS 1-4
    log_likelihood_old = -np.inf
    n, d = X.shape
    means = X[np.random.choice(n, k, replace=False)] 
    covariances = [np.cov(X.T) + 1e-6*np.eye(d) for _ in range(k)]
    priors = [1./k] * k 
    #STEP 5 and 13 (the for loop)
    for _ in range(100): #Limiting run to 100 iterations

        #Calculating log probabilities and responsibilities
        log_probs = np.zeros((X.shape[0], k))
        for i in range(k):
            log_probs[:, i] = multivariate_normal.logpdf(X, mean=means[i], cov=covariances[i], allow_singular=True) + np.log(priors[i])
        
        #Expectation Step (steps 7-8) 
        log_responsibilities = log_probs - logsumexp(log_probs, axis=1)[:, np.newaxis]
        responsibilities = np.exp(log_responsibilities)

        #Maximation Step (steps 9-12) 
        Nk = responsibilities.sum(axis=0)
        means = np.dot(responsibilities.T, X) / Nk[:, np.newaxis]
        covariances = []
        for i in range(k):
            diff = X - means[i]
            covariances.append(np.dot(responsibilities[:, i] * diff.T, diff) / Nk[i] + 1e-6*np.eye(d))
        priors = Nk / n
        
        #Checking for convergence - this is also tied to step 5 and 13
        log_likelihood_new = np.sum(logsumexp(log_probs, axis=1))
        if np.abs(log_likelihood_new - log_likelihood_old) < epsilon:
            break
        log_likelihood_old = log_likelihood_new
    
    return responsibilities, means, covariances

#Instructed to run code several times due to different random initial center
#selections and report clustering with the best NMI score
nmi_scores = []
for iteration in range(3): 
    #Calls EM_algorithm function
    responsibilities, means, covariances = EM_algorithm(X, k, epsilon)

    #Assigns points to clusters
    cluster_sizes = responsibilities.argmax(axis=1)

    #Calculates the NMI score
    nmi_score = normalized_mutual_info_score(y_true, cluster_sizes)

    print("CLUSTERS ITERATION "+str(iteration+1)+":\n")

    #Printing the mean vectors
    for i in range(len(means)):
        print("Mean vector for cluster "+str(i+1)+":")
        print(means[i], "\n")

    #Printing the covariance matrices
    for i in range(len(covariances)):
        print("Covariance matrix for cluster "+str(i+1)+":")
        print(covariances[i], "\n")

    #Printing the cluster sizes
    cluster_sizes = np.bincount(cluster_sizes)
    for i in range(len(cluster_sizes)):
        print("Cluster size of cluster "+str(i+1)+":", cluster_sizes[i])
    print("")

    #Printing nmi score and adding it to list of nmi scores:
    print("Normalized Mutual Information (NMI) score:", nmi_score, "\n")
    nmi_scores.append(nmi_score)

#Printing highest nmi score
highest_score = 0
highest_index = 0
for score in range(len(nmi_scores)):
    if nmi_scores[score] > highest_score:
        highest_score = nmi_scores[score]
        highest_index = score+1
print("Highest NMI score is:", highest_score)
print("This score comes from cluster iteration:", highest_index)
```

```python
#PART 2
import numpy as np
from sklearn import datasets
from sklearn.metrics import normalized_mutual_info_score #CSCI 4390 students instructed to use
from scipy.stats import multivariate_normal #Instructed to use
from scipy.special import logsumexp #Instructed to use

#Loading in the iris dataset
iris = datasets.load_iris()
X = iris.data[:, :4]  #Take only the first four features.
y_true = iris.target

#Initialization of variables (STEPS 1 through 4)
np.random.seed(42)
k = 3  #Number of clusters which we are given
n, d = X.shape
means = X[np.random.choice(n, k, replace=False)] 
covariances = [np.cov(X.T) + 1e-6*np.eye(d) for _ in range(k)]
priors = [1./k] * k 

#Function used in calculating log probabilities - needed for the expectation step
def log_probabilities(X, means, covariances, priors):
    log_probs = np.zeros((X.shape[0], k))
    for i in range(k):
        log_probs[:, i] = multivariate_normal.logpdf(X, mean=means[i], cov=covariances[i], allow_singular=True) + np.log(priors[i])
    return log_probs

#em_step function holds the bulk of the EM algorithm. It conducts code for steps 7-8
#(The expectation step) and steps 9-12 (the Maximization step)
def em_step(X, log_probs):
    #Expectation Step
    log_responsibilities = log_probs - logsumexp(log_probs, axis=1)[:, np.newaxis]
    responsibilities = np.exp(log_responsibilities)

    #Maximation Step
    Nk = responsibilities.sum(axis=0)
    means = np.dot(responsibilities.T, X) / Nk[:, np.newaxis]
    covariances = []
    for i in range(k):
        diff = X - means[i]
        covariances.append(np.dot(responsibilities[:, i] * diff.T, diff) / Nk[i] + 1e-6*np.eye(d))
    priors = Nk / n

    return means, covariances, priors, responsibilities

#Expectation-Maximization (EM) Algorithm
epsilon = 1e-4  #Convergence threshold
log_likelihood_old = -np.inf

#STEP 5 & 13 are the for loop
for _ in range(100): #Limiting run to 100 iterations

    #Calls both functions to execute most of the algorithm
    log_probs = log_probabilities(X, means, covariances, priors)
    means, covariances, priors, responsibilities = em_step(X, log_probs)
    
    #Checking for convergence through use of the em_step function
    log_likelihood_new = np.sum(logsumexp(log_probs, axis=1))
    if np.abs(log_likelihood_new - log_likelihood_old) < epsilon:
        break
    log_likelihood_old = log_likelihood_new

#Assigns points to clusters
cluster_sizes = responsibilities.argmax(axis=1)

#Calculates the NMI score
nmi_score = normalized_mutual_info_score(y_true, cluster_sizes)

#Printing the mean vectors
for i in range(len(means)):
    print("Mean vector for cluster "+str(i+1)+":")
    print(means[i], "\n")

#Printing the covariance matrices
for i in range(len(covariances)):
    print("Covariance matrix for cluster "+str(i+1)+":")
    print(covariances[i], "\n")

#Printing the cluster sizes
cluster_sizes = np.bincount(cluster_sizes)
for i in range(len(cluster_sizes)):
    print("Cluster size of cluster "+str(i+1)+":", cluster_sizes[i])
print("")

#Printing nmi score:
print("Normalized Mutual Information (NMI) score:", nmi_score)
```
